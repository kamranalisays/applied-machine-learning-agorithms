```python
# PCA - Principle Component Analysis
# Dimension reduction
```

```python
# Import libraies
from sklearn.datasets import load_iris
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

import matplotlib.pyplot as plt
```

```python
# Load Data
data = load_iris()
X = pd.DataFrame(data.data , columns= data.feature_names)
print(X.head())
```

```python
# Standardize the data

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Apply PCA 
# Reducing the features from 4 to 2
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
# Check
print(pca.explained_variance_ratio_)
print("Total Variance Captured:" ,sum(pca.explained_variance_ratio_))
```

```python
# That means the first two components capture ~95.8% of the total information in the dataset.
# We can safely ignore the remaining two features.
```

```python


plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0], X_pca[:,1], c=data.target, cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA on Iris Dataset (2D Projection)')
plt.show()
```
